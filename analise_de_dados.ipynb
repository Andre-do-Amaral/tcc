```python
import pandas as pd

colunas_desejadas = ['Data'] + [coluna for coluna in df.columns if coluna.startswith('SC_')]

df_sc = df[colunas_desejadas].copy()
display(df_sc.head())
```

```python

print("Tipos de dados ANTES da conversão:")
print(df_sc.dtypes)

# Percorre cada nome de coluna no DataFrame
for coluna in df_sc.columns:
    # Verifica se a coluna é do tipo 'object'
    if df_sc[coluna].dtype == 'object':
        try:
            # Tenta substituir a vírgula por ponto e converter para float
            df_sc[coluna] = df_sc[coluna].str.replace(',', '.').astype(float)
            print(f"Coluna '{coluna}' convertida para float com sucesso.")
        except AttributeError:
            # Se der erro (ex: a coluna tem valores nulos que não são strings), ignora
            print(f"Coluna '{coluna}' não pôde ser convertida (pode conter não-strings).")
        except ValueError:
            # Se der erro (ex: a coluna é texto de verdade), ignora
            print(f"Coluna '{coluna}' não pôde ser convertida (contém texto não-numérico).")


print("\nTipos de dados DEPOIS da conversão:")
print(df_sc.dtypes)
```

```python
import pandas as pd
import plotly.graph_objects as go

# --- Passo 1: Definição das Variáveis ---
#df_sc['Data'] = pd.to_datetime(df_sc['Data'])
#df_sc.set_index('Data', inplace=True)
dataframe_alvo = df_sc
coluna_volume = 'SC_VolumePorcentagem'
janela_curta = 30
janela_longa = 365
volume_critico = 30.0 # Define o nível crítico em 30%

# --- Passo 2: Cálculo das Médias Móveis ---
# O cálculo é feito diretamente na coluna de interesse do nosso DataFrame.
dataframe_alvo[f'MM_{janela_curta}D'] = dataframe_alvo[coluna_volume].rolling(window=janela_curta).mean()
dataframe_alvo[f'MM_{janela_longa}D'] = dataframe_alvo[coluna_volume].rolling(window=janela_longa).mean()

# --- Passo 3: Criação da Figura e Adição dos Traços ---
fig = go.Figure()

# Adiciona a linha do volume diário original do Sistema Completo
fig.add_trace(go.Scatter(
    x=dataframe_alvo.index, # Usamos o índice do DataFrame para o eixo X
    y=dataframe_alvo[coluna_volume],
    mode='lines',
    name='Volume Diário (SC)',
    line=dict(color='royalblue', width=1.5),
    opacity=0.6
))

# Adiciona a linha da Média Móvel de 30 dias
fig.add_trace(go.Scatter(
    x=dataframe_alvo.index,
    y=dataframe_alvo[f'MM_{janela_curta}D'],
    mode='lines',
    name=f'Média Móvel {janela_curta} Dias (Tendência Mensal)',
    line=dict(color='darkorange', width=2.5)
))

# Adiciona a linha da Média Móvel de 365 dias
fig.add_trace(go.Scatter(
    x=dataframe_alvo.index,
    y=dataframe_alvo[f'MM_{janela_longa}D'],
    mode='lines',
    name=f'Média Móvel {janela_longa} Dias (Tendência Anual)',
    line=dict(color='crimson', width=3)
))

# Adiciona a linha de Nível Crítico
fig.add_hline(
    y=volume_critico,
    line_dash="dot",
    line_color="red",
    line_width=2,
    annotation_text=f"Nível Crítico ({volume_critico}%)",
    annotation_position="bottom right",
    annotation_font=dict(size=12, color="red")
)

# --- Passo 4: Customização do Layout do Gráfico ---
fig.update_layout(
    title_text='Análise de Médias Móveis - Volume útil do Sistema Cantareira',
    xaxis_title='Data',
    yaxis_title='Volume Útil (%)',
    template='plotly_white',
    hovermode='x unified',
    legend=dict(
        orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1
    )
)

fig.show()
```

```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# --- Passo 1: Definição das Variáveis ---
# Assumindo que seu DataFrame se chama df_sc e tem a data como índice.
dataframe_alvo = df_sc
coluna_volume = 'SC_VolumePorcentagem'
coluna_chuva = 'SC_Precipitacao'
janela_mm = 30 # Usaremos uma janela de 30 dias para ambas as médias móveis

# --- Passo 2: Cálculo das Médias Móveis ---
# Criamos as colunas de média móvel para volume e para chuva
dataframe_alvo[f'Volume_MM{janela_mm}D'] = dataframe_alvo[coluna_volume].rolling(window=janela_mm).mean()
dataframe_alvo[f'Chuva_MM{janela_mm}D'] = dataframe_alvo[coluna_chuva].rolling(window=janela_mm).mean()

# --- Passo 3: Criação da Figura com Eixo Y Duplo ---
# Inicializa a figura especificando que teremos um segundo eixo Y
fig = make_subplots(specs=[[{"secondary_y": True}]])

# --- Passo 4: Adicionar os Traços (as linhas do gráfico) ---

# Adiciona a Média Móvel do Volume (Eixo Y da Esquerda)
fig.add_trace(
    go.Scatter(
        x=dataframe_alvo.index,
        y=dataframe_alvo[f'Volume_MM{janela_mm}D'],
        name=f'Média Móvel Volume ({janela_mm} dias)',
        line=dict(color='royalblue', width=2.5)
    ),
    secondary_y=False, # Indica que este é o eixo principal (esquerda)
)

# Adiciona a Média Móvel da Precipitação (Eixo Y da Direita)
fig.add_trace(
    go.Scatter(
        x=dataframe_alvo.index,
        y=dataframe_alvo[f'Chuva_MM{janela_mm}D'],
        name=f'Média Móvel Chuva ({janela_mm} dias)',
        line=dict(color='mediumseagreen', width=2, dash='dot')
    ),
    secondary_y=True, # Indica que este é o eixo secundário (direita)
)

# --- Passo 5: Customização do Layout e dos Eixos ---
fig.update_layout(
    title_text='Volume do Reservatório vs. Precipitação (Médias Móveis de 30 Dias)',
    template='plotly_white',
    hovermode='x unified',
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)
)

# Customiza o eixo X
fig.update_xaxes(title_text="Data")

# Customiza o eixo Y da esquerda (Volume)
fig.update_yaxes(
    title_text="<b>Volume Útil (%)</b>",
    color='royalblue',
    secondary_y=False
)

# Customiza o eixo Y da direita (Precipitação)
fig.update_yaxes(
    title_text="<b>Precipitação Média (mm)</b>",
    color='mediumseagreen',
    secondary_y=True
)

fig.show()
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# --- Assumindo que seu DataFrame é o `df_sc` e a data já é o índice ---

# 1. Criar um DataFrame de análise para garantir que estamos trabalhando a partir de 2000
df_analysis = df_sc.loc['2000':].copy()

# 2. Isolar e limpar a nossa variável alvo
y = df_analysis['SC_VolumePorcentagem'].copy()
y.dropna(inplace=True)

# 3. Criar o DataFrame de features para a tendência
#    A feature 'tempo' representa a passagem do tempo como uma sequência numérica (0, 1, 2...)
X_trend = pd.DataFrame(
    data=dict(tempo=np.arange(len(y))),
    index=y.index
)

# 4. Calcular a tendência manualmente com Regressão Linear
model = LinearRegression()
model.fit(X_trend, y)
y_pred = model.predict(X_trend)

# Criar uma Série pandas com a linha de tendência para facilitar o plot
trend_series = pd.Series(y_pred, index=y.index)

# 5. Plotar o resultado final
plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(16, 8))

# Plota a série temporal original
ax.plot(y.index, y, color='royalblue', alpha=0.6, linewidth=1.5, label='Volume Diário (SC)')

# Plota a linha de tendência calculada
ax.plot(trend_series.index, trend_series, color='crimson', linewidth=3, label='Tendência Linear (2000-Presente)')

# 6. Customizar o gráfico
ax.set_title('Análise de Tendência - Volume do Sistema Cantareira (a partir de 2000)', fontsize=16)
ax.set_xlabel('Data', fontsize=12)
ax.set_ylabel('Volume Útil (%)', fontsize=12)
ax.legend()
plt.show()
```

```python
from statsmodels.tsa.seasonal import seasonal_decompose
import matplotlib.pyplot as plt


# 1. Executar a decomposição
#    - `model='additive'`: Assume que a sazonalidade é somada à tendência. É um bom ponto de partida.
#    - `period=365`: Informa à função que o nosso ciclo principal se repete a cada 365 dias (anual).
decomposicao = seasonal_decompose(y, model='additive', period=365)

# 2. Plotar o resultado
fig = decomposicao.plot()

# 3. Ajustar o tamanho do gráfico para melhor visualização
fig.set_size_inches(16, 10)
plt.show()
```

Os periodos de chuva e seca podem ser confirmados no gráfico a seguir. 

```python
import pandas as pd
import plotly.express as px


# Criamos uma cópia para trabalhar sem alterar o original
df_chuva = df_sc.copy()

# Criamos uma coluna 'Mes' a partir do índice para podermos agrupar
df_chuva['Mes'] = df_chuva.index.month

# Dicionário e lista para mapear e ordenar os meses (do seu código de referência)
nomes_dos_meses = {
    1: 'Jan', 2: 'Fev', 3: 'Mar', 4: 'Abr', 5: 'Mai', 6: 'Jun',
    7: 'Jul', 8: 'Ago', 9: 'Set', 10: 'Out', 11: 'Nov', 12: 'Dez'
}
df_chuva['Nome do Mês'] = df_chuva['Mes'].map(nomes_dos_meses)

ordem_dos_meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 
                   'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez']


# --- Passo 1: Agrupar os dados e calcular a MÉDIA de chuva diária por mês ---
chuva_mensal_media = df_chuva.groupby('Nome do Mês')['SC_Precipitacao'].mean().reset_index()


# --- Passo 2: Criar o gráfico de barras com os dados de média ---
fig = px.bar(
    chuva_mensal_media,
    x='Nome do Mês',
    y='SC_Precipitacao',
    title='Média de Precipitação Diária por Mês no Sistema Cantareira',
    labels={
        'Nome do Mês': 'Mês',
        'SC_Precipitacao': 'Média da Precipitação Diária (mm)'
    },
    template='plotly_white',
    category_orders={"Nome do Mês": ordem_dos_meses}, # Garante a ordem correta dos meses
    text_auto=True
)

# --- Passo 3: Aprimorar a formatação do texto ---
fig.update_traces(
    texttemplate='%{y:.2f} mm', 
    textposition='outside',
    hovertemplate='<b>Mês:</b> %{x}<br><b>Média Diária:</b> %{y:.2f} mm<extra></extra>'
)

# --- Passo 4: Exibir o gráfico ---
fig.show()
```

Análise de autocorrelação (lags)

```python
from pandas.plotting import lag_plot
import matplotlib.pyplot as plt

residuos = decomposicao.resid

# A decomposição cria valores nulos no início e no fim da série, vamos removê-los
residuos.dropna(inplace=True)
# Cria o lag plot para os resíduos
lag_plot(residuos)

plt.title('Lag Plot dos Resíduos (Hoje vs. Ontem)')
plt.xlabel('Resíduo de Ontem')
plt.ylabel('Resíduo de Hoje')
plt.show()
```

```python
from statsmodels.graphics.tsaplots import plot_pacf
import matplotlib.pyplot as plt

# --- Assumindo que a série `residuos` já foi calculada e está na memória ---

# 1. Criar uma figura com um único subplot, com um bom tamanho
fig, ax = plt.subplots(figsize=(12, 6))

# 2. Plotar o PACF, especificando o número de lags que queremos ver (ex: 40)
plot_pacf(residuos, ax=ax, lags=40)

# 3. Customizar o gráfico
ax.set_title('Função de Autocorrelação Parcial (PACF) dos Resíduos', fontsize=16)
ax.set_xlabel('Número do Lag', fontsize=12)
ax.set_ylabel('Correlação Parcial', fontsize=12)
ax.grid(True, linestyle='--', alpha=0.6)

plt.show()
```
