```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# --- Assumindo que seu DataFrame é o `df_sc` e a data já é o índice ---

# 1. Criar um DataFrame de análise para garantir que estamos trabalhando a partir de 2000
df_analysis = df_sc.loc['2000':].copy()

# 2. Isolar e limpar a nossa variável alvo
y = df_analysis['SC_VolumeOperacionalHm3'].copy()
y.dropna(inplace=True)

# 3. Criar o DataFrame de features para a tendência
#    A feature 'tempo' representa a passagem do tempo como uma sequência numérica (0, 1, 2...)
X_trend = pd.DataFrame(
    data=dict(tempo=np.arange(len(y))),
    index=y.index
)

# 4. Calcular a tendência manualmente com Regressão Linear
model = LinearRegression()
model.fit(X_trend, y)
y_pred = model.predict(X_trend)

# Criar uma Série pandas com a linha de tendência para facilitar o plot
trend_series = pd.Series(y_pred, index=y.index)

# 5. Plotar o resultado final
plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(16, 8))

# Plota a série temporal original
ax.plot(y.index, y, color='royalblue', alpha=0.6, linewidth=1.5, label='Volume Diário (SC)')

# Plota a linha de tendência calculada
ax.plot(trend_series.index, trend_series, color='crimson', linewidth=3, label='Tendência Linear (2000-Presente)')

# 6. Customizar o gráfico
ax.set_title('Análise de Tendência - Volume do Sistema Cantareira (a partir de 2000)', fontsize=16)
ax.set_xlabel('Data', fontsize=12)
ax.set_ylabel('Volume Útil (%)', fontsize=12)
ax.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import periodogram

def plot_periodogram(ts, detrend='linear', ax=None):
    """
    Plota o periodograma de uma série temporal.
    O eixo X é formatado para mostrar frequências anuais (Anual, Mensal, etc.).
    """
    frequencies, spectrum = periodogram(
        ts,
        fs=365,
        detrend=detrend,
        window="boxcar",
        scaling='spectrum',
    )
    
    if ax is None:
        _, ax = plt.subplots(figsize=(12, 7))
        
    ax.step(frequencies, spectrum, color="purple")
    ax.set_xscale("log")
    
    # Define os marcadores do eixo X para os ciclos mais comuns
    ax.set_xticks([1, 2, 4, 6, 12, 26, 52])
    ax.set_xticklabels(
        [
            "Anual (1)",
            "Semestral (2)",
            "Trimestral (4)",
            "Bimestral (6)",
            "Mensal (12)",
            "Quinzenal (26)",
            "Semanal (52)",
        ],
        rotation=30,
    )
    
    ax.set_ylabel("Potência / Variância")
    ax.set_xlabel("Frequência (Ciclos por Ano)")
    ax.set_title("Periodograma", fontsize=16)
    
    return ax


```

```python
plot_periodogram(y)
plt.show()
```

Analisando o periodograma é possivel confirmar a sazonalidade anual. Porém o gráfico ficou estranho a esquerda, parece haver uma sazonalidade com um periodo maior que o anual, pode ser decorrente do comportamento de cheias e baixas estremas em certos periodos, como nas crises hidricas de 2004 e 2014. 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import periodogram

def plot_periodogram_longo(ts, detrend='linear', ax=None):
    """
    Plota o periodograma de uma série temporal, com foco em ciclos
    de curto e longo prazo (incluindo ciclos de múltiplos anos).
    """
    
    frequencies, spectrum = periodogram(
        ts,
        fs=365,
        detrend=detrend,
        window="boxcar",
        scaling='spectrum',
    )
    
    if ax is None:
        _, ax = plt.subplots(figsize=(12, 7))
        
    ax.step(frequencies, spectrum, color="purple")
    ax.set_xscale("log")
    
    ax.set_xticks([0.1, 0.2, 0.5, 1, 2, 4, 6, 12, 26, 52])
    ax.set_xticklabels(
        [
            "Decenal (0.1)",     
            "5 Anos (0.2)",      
            "Bienal (0.5)",     
            "Anual (1)",
            "Semestral (2)",
            "Trimestral (4)",
            "Bimestral (6)",
            "Mensal (12)",
            "Quinzenal (26)",
            "Semanal (52)",
        ],
        rotation=30,
        ha='right'
    )
    
    ax.set_ylabel("Potência / Variância")
    ax.set_xlabel("Frequência (Ciclos por Ano)")
    ax.set_title("Periodograma com Ciclos de Longo Prazo", fontsize=16)
    
    return ax
```

```python
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
import matplotlib.pyplot as plt

# Criar uma figura com dois subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 7))

# Plotar o ACF (Função de Autocorrelação)
plot_acf(residuos, lags=40, ax=ax1)
ax1.set_title('Função de Autocorrelação (ACF)')

# Plotar o PACF (Função de Autocorrelação Parcial)
plot_pacf(residuos, lags=40, ax=ax2)
ax2.set_title('Função de Autocorrelação Parcial (PACF)')

plt.show()
```

```python
from pandas.plotting import lag_plot
import matplotlib.pyplot as plt


fig, axes = plt.subplots(3, 3, figsize=(15, 15))

fig.suptitle('Análise Visual de Lags para os Resíduos', fontsize=18)

for i in range(1, 10):

    # axes.flatten() transforma a matriz 3x2 em uma lista simples para facilitar o acesso
    ax = axes.flatten()[i-1]
    lag_plot(residuos, lag=i, ax=ax, s=10, alpha=0.5) 
    ax.set_title(f'Lag {i}')
    ax.grid(True)


plt.tight_layout(rect=[0, 0.03, 1, 0.95]) 

plt.show()
```

```python
import pandas as pd
from statsmodels.tsa.deterministic import DeterministicProcess, CalendarFourier


# Componente de Sazonalidade Anual com Termos de Fourier
# freq='A' significa Anual. order=4 cria 8 features (4 senos, 4 cossenos)
fourier = CalendarFourier(freq='A', order=1)

#    - index: O índice de tempo da nossa série y.
#    - constant=True: Adiciona o intercepto (termo constante) ao modelo.
#    - order=3: Cria features de tendência (linear, quadrática e cúbica).
#               -> Você pode alterar este número para testar outras ordens!
#    - additional_terms=[fourier]: Adiciona nossas features de sazonalidade.
#    - drop=True: Remove a primeira feature de cada tipo para evitar multicolinearidade.
dp = DeterministicProcess(
    index=y.index,
    constant=True,
    order=3,
    additional_terms=[fourier],
    drop=True,
)

X1 = dp.in_sample()

print("DataFrame de features X1 para o Modelo Linear criado com sucesso!")
print(f"Número de features criadas: {X1.shape[1]}")
print("\nPrimeiras 5 linhas de X1:")
display(X1.head())

print("\nÚltimas 5 linhas de X1:")
display(X1.tail())
```

```python
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import pandas as pd


model_1 = LinearRegression()
model_1.fit(X1, y)

y_pred_1 = model_1.predict(X1)

y_pred_1 = pd.Series(y_pred_1, index=y.index)

# Plotar o resultado para avaliar o ajuste do modelo
plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(18, 9))

# Plota a série temporal original
ax.plot(y.index, y, color='royalblue', alpha=0.6, linewidth=1.5, label='Volume Real Diário')

# Plota a previsão do nosso Modelo 1 (Tendência + Sazonalidade)
ax.plot(y_pred_1.index, y_pred_1, color='crimson', linewidth=2.5, label='Previsão do Modelo Linear (Tendência + Sazonalidade)')

ax.set_title('Ajuste do Modelo Linear (Tendência + Sazonalidade)', fontsize=18)
ax.set_xlabel('Data', fontsize=14)
ax.set_ylabel('Volume Operacional (hm³)', fontsize=14)
ax.legend(fontsize=12)
plt.show()
```

O gráfico acima mostra a previsão da Linear Regression com as features de tendência e sazonalidade, elas não capturam os picos dos dados, por isso devemos calcular os residuos, que serão utilizados no modelo XGBoost.

```python
import pandas as pd
import matplotlib.pyplot as plt

y_residuos = y - y_pred_1
y_residuos.name = 'residuos_volume'

fig, ax = plt.subplots(figsize=(18, 7))
ax.plot(y_residuos.index, y_residuos, color='indigo', alpha=0.8, linewidth=1)
ax.axhline(0, color='red', linestyle='--', linewidth=2) # Adiciona uma linha no zero para referência

ax.set_title('Resíduos do Modelo Linear (Tendência + Sazonalidade)', fontsize=16)
ax.set_xlabel('Data', fontsize=12)
ax.set_ylabel('Erro de Previsão (hm³)', fontsize=12)
plt.show()
```

```python



X2 = pd.DataFrame()


#    O método .shift(i) "puxa" os valores de `i` dias atrás para a linha atual.
for i in range(1, 10): 
    X2[f'lag_{i}'] = y_residuos.shift(i)


#    O método .shift() cria NaNs no início do DataFrame. Precisamos removê-los.
X2.dropna(inplace=True)

# 4. ALINHAMENTO CRÍTICO: Garantir que `y_residuos` e `X2` tenham o mesmo tamanho e índice
#    Como removemos linhas de X2, precisamos fazer o mesmo em y_residuos.
y_residuos = y_residuos.reindex(X2.index)



print("DataFrame de features X2 para o Modelo de Resíduos criado com sucesso!")
print("\nPrimeiras 5 linhas de X2 (features de lag):")
display(X2.head())

print(f"\nTamanho de X2: {X2.shape}")
print(f"Tamanho de y_residuos: {y_residuos.shape}")
```

```python
import pandas as pd


#    Isso garante que todas as nossas matrizes de dados tenham exatamente as mesmas datas.
y_final = y.reindex(X2.index)
X1_final = X1.reindex(X2.index)
y_residuos_final = y_residuos.reindex(X2.index) # Já estava alinhado, mas é bom garantir.
X2_final = X2.copy() # Já é a nossa referência


cutoff_date = '2024-01-01'

#Criar uma máscara booleana para treino e teste
mask_train = X1_final.index < cutoff_date
mask_test = X1_final.index >= cutoff_date

X1_train, X1_test = X1_final[mask_train], X1_final[mask_test]
X2_train, X2_test = X2_final[mask_train], X2_final[mask_test]
y_train, y_test = y_final[mask_train], y_final[mask_test]
y_residuos_train, y_residuos_test = y_residuos_final[mask_train], y_residuos_final[mask_test]


print("Divisão dos dados concluída com sucesso!")
print("--- CONJUNTO DE TREINO ---")
print(f"Shape de X1_train: {X1_train.shape}")
print(f"Shape de X2_train: {X2_train.shape}")
print(f"Shape de y_train: {y_train.shape}")
print(f"Período de treino: de {X1_train.index.min().date()} até {X1_train.index.max().date()}")

print("\n--- CONJUNTO DE TESTE ---")
print(f"Shape de X1_test: {X1_test.shape}")
print(f"Shape de X2_test: {X2_test.shape}")
print(f"Shape de y_test: {y_test.shape}")
print(f"Período de teste: de {X1_test.index.min().date()} até {X1_test.index.max().date()}")
```
