```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime, timedelta, time, date
import plotly.express as px

def to_time(x):
    return datetime.strptime(x.replace("-","/"), '%Y/%m/%d %H:%M:%S')

def remove_unit(x):
    if "%" in x:
        return float(x.split("%")[0])
    else:
        return float(x.split()[0])

def to_inteiro(x):
    return int(x)

def isNaN(x):
    return x != x


def to_month(x):
    return x.month


def to_day_(x):
    hour = x.hour
    minutes = x.minute
    seconds = x.second
    return x - timedelta(hours=hour, minutes = minutes, seconds = seconds)

def to_day(x):
    return x.day

def to_week(x):
    if x.year == 2021:
        if x.week == 52:
            if x.month == 1:
                return timedelta(1 * 7) + datetime.strptime("2021-01-01 00:00:00".replace("-","/"), '%Y/%m/%d %H:%M:%S')
            elif x.month == 12:
                return timedelta(x.week * 7) + datetime.strptime("2021-01-01 00:00:00".replace("-","/"), '%Y/%m/%d %H:%M:%S')
        else:
            return timedelta(x.week * 7) + datetime.strptime("2021-01-01 00:00:00".replace("-","/"), '%Y/%m/%d %H:%M:%S')
    elif x.year == 2022:
        if x.week == 52:
            if x.month == 1:
                return timedelta(1 * 7) + datetime.strptime("2022-01-01 00:00:00".replace("-","/"), '%Y/%m/%d %H:%M:%S')
            elif x.month == 12:
                return timedelta(x.week * 7) + datetime.strptime("2022-01-01 00:00:00".replace("-","/"), '%Y/%m/%d %H:%M:%S')
        else:
            return timedelta(x.week * 7) + datetime.strptime("2022-01-01 00:00:00".replace("-","/"), '%Y/%m/%d %H:%M:%S')
    elif x.year == 2020:
        if x.week == 52:
            if x.month == 1:
                return timedelta(1 * 7) + datetime.strptime("2020-01-01 00:00:00".replace("-","/"), '%Y/%m/%d %H:%M:%S')
            elif x.month == 12:
                return timedelta(x.week * 7) + datetime.strptime("2020-01-01 00:00:00".replace("-","/"), '%Y/%m/%d %H:%M:%S')
        else:
            return timedelta(x.week * 7) + datetime.strptime("2020-01-01 00:00:00".replace("-","/"), '%Y/%m/%d %H:%M:%S')
    elif x.year == 2023:
        if x.week == 52:
            if x.month == 1:
                return timedelta(1 * 7) + datetime.strptime("2023-01-01 00:00:00".replace("-","/"), '%Y/%m/%d %H:%M:%S')
            elif x.month == 12:
                return timedelta(x.week * 7) + datetime.strptime("2023-01-01 00:00:00".replace("-","/"), '%Y/%m/%d %H:%M:%S')
        else:
            return timedelta(x.week * 7) + datetime.strptime("2023-01-01 00:00:00".replace("-","/"), '%Y/%m/%d %H:%M:%S')
    elif x.year == 2024:
        if x.week == 52:
            if x.month == 1:
                return timedelta(1 * 7) + datetime.strptime("2024-01-01 00:00:00".replace("-","/"), '%Y/%m/%d %H:%M:%S')
            elif x.month == 12:
                return timedelta(x.week * 7) + datetime.strptime("2024-01-01 00:00:00".replace("-","/"), '%Y/%m/%d %H:%M:%S')
        else:
            return timedelta(x.week * 7) + datetime.strptime("2024-01-01 00:00:00".replace("-","/"), '%Y/%m/%d %H:%M:%S')

def to_time_2(x):
            return datetime.strptime(x.replace("-","/"), '%m/%d/%Y %H:%M')

def ajuste_time(x):
            hora = x.hour
            minuto = x.minute
            minutos_tot = hora*60 + minuto
            if (0<=minutos_tot)&(minutos_tot<90):
                day = x.day
                month = x.month
                year = x.year
                data_ = str(day) + "/" + str(month) + "/" + str(year) + " 00:00"
                return datetime.strptime(data_, '%d/%m/%Y %H:%M')
            elif (90<=minutos_tot)&(minutos_tot<270):
                day = x.day
                month = x.month
                year = x.year
                data_ = str(day) + "/" + str(month) + "/" + str(year) + " 03:00"
                return datetime.strptime(data_, '%d/%m/%Y %H:%M')
            elif (270<=minutos_tot)&(minutos_tot<450):
                day = x.day
                month = x.month
                year = x.year
                data_ = str(day) + "/" + str(month) + "/" + str(year) + " 06:00"
                return datetime.strptime(data_, '%d/%m/%Y %H:%M')
            elif (450<=minutos_tot)&(minutos_tot<630):
                day = x.day
                month = x.month
                year = x.year
                data_ = str(day) + "/" + str(month) + "/" + str(year) + " 09:00"
                return datetime.strptime(data_, '%d/%m/%Y %H:%M')
            elif (630<=minutos_tot)&(minutos_tot<810):
                day = x.day
                month = x.month
                year = x.year
                data_ = str(day) + "/" + str(month) + "/" + str(year) + " 12:00"
                return datetime.strptime(data_, '%d/%m/%Y %H:%M')
            elif (810<=minutos_tot)&(minutos_tot<990):
                day = x.day
                month = x.month
                year = x.year
                data_ = str(day) + "/" + str(month) + "/" + str(year) + " 15:00"
                return datetime.strptime(data_, '%d/%m/%Y %H:%M')
            elif (990<=minutos_tot)&(minutos_tot<1170):
                day = x.day
                month = x.month
                year = x.year
                data_ = str(day) + "/" + str(month) + "/" + str(year) + " 18:00"
                return datetime.strptime(data_, '%d/%m/%Y %H:%M')
            elif (1170<=minutos_tot)&(minutos_tot<1441):
                day = x.day
                month = x.month
                year = x.year
                data_ = str(day) + "/" + str(month) + "/" + str(year) + " 21:00"
                return datetime.strptime(data_, '%d/%m/%Y %H:%M')

# FUNÇÂO INVES DO MERGE
def cafe(x):
    try:
        return dad[dad["bebida"]==x]["café"].iloc[0]
    except:
        return 0
def gelado(x):
    try:
        return dad[dad["bebida"]==x]["gelado"].iloc[0]
    except:
        return 0

def leite(x):
    try:
        return dad[dad["bebida"]==x]["leite"].iloc[0]
    except:
        return 0
```

```python
def initial(train, test, count):
    # inicializando e setando os parâmetros

    s = setup(data = train[count], test_data = test[count], target = "Flat Caramel",
              fold_strategy = "timeseries",
              numeric_features = ["Series", "Temperatura"],
              fold = 3,
              transform_target = True, session_id = 123)
    # Poderia normalizar os dados usando o normalize = True ou
    # Transformar os dados usando o transform = True
    # no caso eu transformei apenas a distribuição da target

def initial_cf(train, test, count):
    # inicializando e setando os parâmetros

    s = setup(data = train[count], test_data = test[count], target = "coffee",
              fold_strategy = "timeseries",
              numeric_features = ["Series","dayofweek"],
              fold = 3,
              transform_target = True, session_id = 123)
    # Poderia normalizar os dados usando o normalize = True ou
    # Transformar os dados usando o transform = True
    # no caso eu transformei apenas a distribuição da target
```

```python
from sklearn.metrics import make_scorer
def score_fun(y, y_pred):
    return y.sum() - y_pred.sum()

def results(count, train, test):
    initial(train, test, count = count)
    resultados = []
    modelos = compare_models(sort="MAE", n_select = 10)
    pos = []
    countt = 0
    modelo_result = []
    compara = np.array(test[count]["Flat Caramel"])
    soma = []
    for model in modelos:
        resultados.append(score_fun(np.array(test[count]["Flat Caramel"]),np.array(predict_model(model, test[count])["Label"])))
        pos.append(f'{countt}')
        countt = countt + 1
        modelo_result.append(np.array(predict_model(model, test[count])["Label"]))
        soma.append(np.array(predict_model(model, test[count])["Label"]).sum())
    soma.append(np.array(test[count]["Flat Caramel"]).sum())
    soma = pd.DataFrame(soma, columns = [10])
    pos = pd.Series(pos, name = "Posição Modelos")
    resultados = pd.Series(resultados, name = "Diff 21 dias")
    resultados = pd.concat([pos, resultados], axis = 1)
    resultados["ABS Diff 21 dias"] = resultados["Diff 21 dias"].apply(abs)
    resultados = resultados.sort_values("ABS Diff 21 dias")
    prev = pd.DataFrame(modelo_result).T
    prev = pd.concat([prev, pd.Series(compara, name = 10)], axis = 1)
    prev.set_index(test[count].index.date, inplace = True)
    prev = pd.concat([prev, soma.T]).rename(columns = {10:"REAL"})
    return resultados, prev, modelos


def results_cf(count, train, test):
    initial_cf(train, test, count = count)
    resultados = []
    modelos = compare_models(sort="MAE", n_select = 10)
    pos = []
    countt = 0
    modelo_result = []
    compara = np.array(test[count]["coffee"])
    soma = []
    for model in modelos:
        resultados.append(score_fun(np.array(test[count]["coffee"]),np.array(predict_model(model, test[count])["Label"])))
        pos.append(f'{countt}')
        countt = countt + 1
        modelo_result.append(np.array(predict_model(model, test[count])["Label"]))
        soma.append(np.array(predict_model(model, test[count])["Label"]).sum())
    soma.append(np.array(test[count]["coffee"]).sum())
    soma = pd.DataFrame(soma, columns = [10])
    pos = pd.Series(pos, name = "Posição Modelos")
    resultados = pd.Series(resultados, name = "Diff 21 dias")
    resultados = pd.concat([pos, resultados], axis = 1)
    resultados["ABS Diff 21 dias"] = resultados["Diff 21 dias"].apply(abs)
    resultados = resultados.sort_values("ABS Diff 21 dias")
    prev = pd.DataFrame(modelo_result).T
    prev = pd.concat([prev, pd.Series(compara, name = 10)], axis = 1)
    prev.set_index(test[count].index.date, inplace = True)
    prev = pd.concat([prev, soma.T]).rename(columns = {10:"REAL"})
    return resultados, prev, modelos
```
